**Once we have our library installed, its time to import the required tools into Python.**

```python
from datetime import date
from icrawler.builtin import BaiduImageCrawler
from google.colab import drive
from pathlib import Path
import os.path
from os import path
```

**Before we start scraping, we need to define the folders that we will be storing images in.**

```python
if path.exists('/content/images') == False:
  os.mkdir('/content/images')
  os.mkdir('/content/images/fish_images')
  os.mkdir('/content/images/bird_images')
  os.mkdir('/content/images/dog_images')
  os.mkdir('/content/images/cat_images')
root = Path.cwd()/"images"
```

**The code below allows us to scrape Baidu Images. You can edit the keyword parameter to change what you are searching for, and adjust max_num to set the exact number of images you want to download.**

```python
baidu_crawler = BaiduImageCrawler(feeder_threads=1, parser_threads=2, downloader_threads=4, storage={'root_dir': '/content/images/fish_images'})

baidu_crawler.crawl(keyword='fishes', filters=None, max_num=50, file_idx_offset='auto')
```
